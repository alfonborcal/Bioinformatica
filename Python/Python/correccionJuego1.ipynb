```python
while True:
    menu = input("Escribe 1 para empezar o 2 para salir")

    if menu == "1":
        print("INICIO")
        HP=100
        SP=50
        print(f"De inicio tienes {HP} de HP y {SP} de SP.")
        
        print("Tienes dos opciones: 1 para pasillo / 2 para palanca")
        opcion = input("Pon tu elección")

        menuBool = True
        while menuBool:
            if opcion == "1":
                print("pasillo")
                menuBool = False
                break
            elif opcion == "2":
                print("palanca")
                menuBool = False
                break
            else:
                print("Escribiste mal")
                opcion = input("Pon tu elección")
    elif menu == "2":
        print("CERRANDO JUEGO")
        break
    else:
        print("No sé que dices")
```
