```python
while True:
    menu = input("Escribe 1 para empezar o 2 para salir")

    if menu == "1":
        print("INICIO")
        HP=100
        SP=100
        print(f"De inicio tienes {HP} de HP y {SP} de SP.")
        
        print("Tienes dos opciones: 1 para pasillo / 2 para palanca")
        opcion1 = input("Pon tu elección")

        eleccion1 = True
        while eleccion1:
            if opcion1 == "1":
                print("pasillo")
                eleccion1 = False
                break
            elif opcion1 == "2":
                print("palanca")
                eleccion1 = False
                break
            else:
                print("Escribiste mal")
                opcion1 = input("Pon tu elección")
    elif menu == "2":
        print("CERRANDO JUEGO")
        break
    else:
        print("No sé que dices")
```
